# KG-DBN

Notebook eksplorasi. Logika pipeline ada di package `kgdbn/`; lihat `docs/ARCHITECTURE.md`.

Alur: ontologi → knowledge graph → embedding node2vec → fitur kasus → DBN → evaluasi.

```python
from kgdbn import DBN, kg_features, load_cases, load_ontology, multi_hot, node2vec
from kgdbn.experiment import DEFAULT_ONTOLOGY, run

CASES = "data/kasus.csv"  # kolom: gejala (dipisah ;), label
```

## 1. Knowledge graph dari ontologi

```python
kg = load_ontology(DEFAULT_ONTOLOGY)
print(kg.summary())
kg.dropped
```

## 2. Embedding KG

```python
embeddings = node2vec(kg, dim=32)
len(embeddings), next(iter(embeddings.values())).shape
```

## 3. Data kasus

```python
cases = load_cases(CASES, kg)
cases["label"].value_counts()
```

## 4. Eksperimen lengkap (semua fitur × semua model)

```python
results = run(CASES)
results.pivot(index="model", columns="fitur", values="f1_macro")
```
